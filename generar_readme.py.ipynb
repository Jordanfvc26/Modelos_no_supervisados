```python
import os

# Carpeta donde están las imágenes
carpeta_imagenes = "visualizaciones"

# Lista de imágenes (puedes mantenerla fija o leer automáticamente)
imagenes = [
    "clusters_dbscan_original.png",
    "clusters_kmeans_original.png",
    "dbscan_eps_selection.png",
    "elbow_method.png",
    "ingreso_vs_gasto.png",
    "pca_clusters_comparacion.png",
    "perfiles_comparativos.png",
    "silhouette_scores.png",
    "tsne_clusters_comparacion.png"
]

# Mapeo de nombres a títulos más legibles (opcional, pero recomendado)
titulos = {
    "clusters_dbscan_original.png": "Clusters detectados por DBSCAN",
    "clusters_kmeans_original.png": "Clusters detectados por K-Means",
    "dbscan_eps_selection.png": "Selección de eps (K-Distance Graph)",
    "elbow_method.png": "Método del codo para K-Means",
    "ingreso_vs_gasto.png": "Relación Ingreso vs Gasto (datos originales)",
    "pca_clusters_comparacion.png": "Reducción con PCA - Clusters K-Means",
    "perfiles_comparativos.png": "Comparativa de perfiles por cluster",
    "silhouette_scores.png": "Puntuaciones Silhouette para diferentes k",
    "tsne_clusters_comparacion.png": "Reducción con t-SNE - Clusters K-Means"
}

# Generar el contenido del README.md
with open("README.md", "w", encoding="utf-8") as f:
    f.write("# 📊 Visualizaciones del Proyecto\n\n")
    f.write("Este documento contiene todos los gráficos generados en el notebook de segmentación.\n\n")
    f.write("## 📈 Listado de gráficos\n\n")
    
    for img in imagenes:
        ruta_completa = os.path.join(carpeta_imagenes, img)
        # Si existe un título personalizado, lo uso; si no, uso el nombre del archivo
        titulo = titulos.get(img, img.replace("_", " ").replace(".png", "").title())
        f.write(f"### {titulo}\n\n")
        f.write(f"![{titulo}]({ruta_completa})\n\n")
        f.write("---\n\n")  # Separador entre gráficos

    f.write("## 📌 Nota\n\n")
    f.write("Los gráficos han sido generados automáticamente desde el script `generar_readme.py`.\n")

print("✅ README.md generado correctamente con todas las imágenes.")
```
